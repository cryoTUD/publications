```python
from locscale.include.emmer.ndimage.map_utils import load_map, save_as_mrc 
from locscale.include.emmer.ndimage.filter import low_pass_filter
```

```python
import os 
emmap_path = "/home/abharadwaj1/thesis/data_archive/inputs/3_surfer/low_pass_filtered/emd_3061/EMD_3061_unfiltered.mrc"
emmap, apix = load_map(emmap_path)
cutoff_res = [5, 6, 7, 8, 9, 10, 15, 20, 25]
for res in cutoff_res:
    print(f"Applying low-pass filter with cutoff resolution: {res} Å")
    new_fold = os.path.join(os.path.dirname(emmap_path), f"low_pass_{res}A")
    os.makedirs(new_fold, exist_ok=True)
    filtered_emmap = low_pass_filter(emmap, apix=apix, cutoff=res)
    output_path = os.path.join(new_fold, f"emd_3061_filtered_{res}A.mrc")
    save_as_mrc(filtered_emmap, output_path, apix=apix)
```

```python
pvddt_map_path = "/home/abharadwaj1/thesis/data_archive/inputs/3_surfer/low_pass_filtered/emd_33888/hallucinations_probabilities_map.mrc"
pvddt_emmap, pvddt_apix = load_map(pvddt_map_path)
pvddt_emmap_invert = pvddt_emmap * -1 
save_as_mrc(pvddt_emmap_invert, pvddt_map_path.replace("hallucinations_probabilities_map.mrc", "emd_33888_pvddt.mrc"), apix=pvddt_apix)
```
